```julia
using CSV, DataFrames, Plots
```

```julia
file_path = "../data/dataModelCSV/filtered_data_elpino.csv"
filter_model1 = CSV.read(file_path, DataFrame; missingstring=["NA", "NULL", "-"], delim=';')
first(filter_model1, 3)
```

```julia
nombres_columnas = names(filter_model1)
```

```julia
# Define el valor de reemplazo
valor_reemplazo = "?00-Posiblemente Recuperado"

# Busca todas las columnas que contienen "Diag" y "Secundario"
columnas_diag = [nombre for nombre in names(filter_model1) 
                if occursin("Diag", nombre) && occursin("Secundario", nombre)]

# Realiza el reemplazo para missing, strings vacíos o nothing
for col in columnas_diag
    for i in 1:nrow(filter_model1)
        valor = filter_model1[i, col]
        # Reemplazar si es missing, string vacío o nothing
        if ismissing(valor) || valor == "" || valor === nothing || (isa(valor, String) && strip(valor) == "")
            filter_model1[i, col] = valor_reemplazo
        end
    end
end

# Muestra las primeras filas para verificar
first(filter_model1, 3)
```
