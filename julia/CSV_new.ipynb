# Rellenar los valores en las columnas de Diag Secudario 

```julia
# Define el valor de reemplazo
valor_reemplazo = "?00-Posiblemente Recuperado"

# Busca todas las columnas que contienen "Diag" y "Secundario"
columnas_diag = [nombre for nombre in names(filter_model1) 
                if occursin("Diag", nombre) && occursin("Secundario", nombre)]

# Realiza el reemplazo para missing, strings vacíos o nothing
for col in columnas_diag
    for i in 1:nrow(filter_model1)
        valor = filter_model1[i, col]
        # Reemplazar si es missing, string vacío o nothing
        if ismissing(valor) || valor == "" || valor === nothing || (isa(valor, String) && strip(valor) == "")
            filter_model1[i, col] = valor_reemplazo
        end
    end
end

# Muestra las primeras filas para verificar
first(filter_model1, 3)
```

```julia
# Define el valor de reemplazo
valor_reemplazo = "0000.00 - Ya no necesita procedimiento"

# Busca todas las columnas que contienen "Diag" y "Secundario"
columnas_diag = [nombre for nombre in names(filter_model1) 
                if occursin("Proced", nombre) && occursin("Secundario", nombre)]

# Realiza el reemplazo para missing, strings vacíos o nothing
for col in columnas_diag
    for i in 1:nrow(filter_model1)
        valor = filter_model1[i, col]
        # Reemplazar si es missing, string vacío o nothing
        if ismissing(valor) || valor == "" || valor === nothing || (isa(valor, String) && strip(valor) == "")
            filter_model1[i, col] = valor_reemplazo
        end
    end
end

# Muestra las primeras filas para verificar
first(filter_model1, 3)
```

```julia
function is_vacio(x)
    if ismissing(x) || x == nothing || x == ""
        return true
    elseif isa(x, Number) && isnan(x)
        return true
    else
        return false
    end
end

function count_vacios_as_dataframe(df::DataFrame)::DataFrame
    columnas = names(df)
    conteo = [count(is_vacio, df[!, col]) for col in columnas]
    return DataFrame(Columna=columnas, Vacios=conteo)
end
conteo_vacios = count_vacios_as_dataframe(filter_model1)
```

# Show only code in Diag

```julia
# Procesar todas las columnas de diagnóstico (del 1 al 35)
for i in 1:35
    column_name = i == 1 ? "Diag 01 Principal (cod+des)" : "Diag $(lpad(i, 2, '0')) Secundario (cod+des)"
    
    # Verificar si la columna existe en el DataFrame
    if column_name in names(data_model2)
        # Extraer solo el código (parte antes del primer "-")
        data_model2[!, column_name] = map(data_model2[!, column_name]) do value
            if ismissing(value) || isnothing(value) || value == ""
                return missing
            else
                # Dividir por el primer "-" y tomar la primera parte
                parts = split(value, "-", limit=2)
                return length(parts) > 0 ? strip(parts[1]) : missing
            end
        end
    end
end
```
