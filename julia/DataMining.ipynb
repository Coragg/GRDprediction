# Analyze data

## Gender data

```julia
column_sex = getproperty(df, :"Sexo (Desc)")
unique_genders = unique(column_sex)
formatted_genders = ["$value" for value in unique_genders]

# Imprimir las cadenas separadas por saltos de línea
println(join(formatted_genders, "\n"))
```

```julia
# Contar la cantidad de "Hombre" y "Mujer"
count_hombre = count(==(formatted_genders[1]), column_sex)
count_mujer = count(==(formatted_genders[2]), column_sex)
gender_counts = [count_hombre, count_mujer]
```

```julia
# Calcular porcentajes
total = sum(gender_counts)
percentages = [count / total * 100 for count in gender_counts]

# Crear etiquetas con los valores numéricos
labels = [string(round(percentages[i], digits=2), "%") for i in 1:length(formatted_genders)]

# Graficar pastel
p = pie(formatted_genders,gender_counts, title="Men and Women")

# Agregar anotaciones dentro de cada sector
theta = cumsum([0; gender_counts]) ./ total .* 2π  # Calcular ángulos para las etiquetas
for i in 1:length(gender_counts)
    x = 0.6 * cos((theta[i] + theta[i+1]) / 2)  
    y = 0.6 * sin((theta[i] + theta[i+1]) / 2) 
   annotate!(x, y, text(labels[i], :black, 10, :bold))
end

display(p)
```

## GRD

```julia
#println(df[:,:GRD])
column_grd = df.GRD  
grd_unico = unique(column_grd)
grd_list = ["$value" for value in grd_unico]  

println("This list has $(length(grd_unico)) GRD, and all of them are:")
for (i, each_grd) in enumerate(grd_list)
    println("    $(i)) $each_grd")
end
```

```julia
# Contar ocurrencias de cada GRD
grd_counts = countmap(df.GRD)

# Obtener los 20 GRD más frecuentes
top_20_grd = sort(collect(grd_counts), by=x -> -x[2])[1:min(20, length(grd_counts))]

# Obtener los 20 GRD menos frecuentes
bottom_20_grd = sort(collect(grd_counts), by=x -> x[2])[1:min(20, length(grd_counts))]

println("Top 20 most frequent GRD:")
for (i, (grd, count)) in enumerate(top_20_grd)
    println("    $(i)) $grd -> $count times")
end
```

```julia
partos = 813 + 639 + 538 + 389 + 325 + 187 + 140
porcentaje_partos = round(partos / rows * 100, digits=2)
println("Este es un aproximado de $(porcentaje_partos)% de los datos de parto en el dataset.")
```

```julia
println("\nTop 20 least frequent GRD:")
for (i, (grd, count)) in enumerate(bottom_20_grd)
    println("    $(i)) $grd -> $count times")
end
```

```julia
println("\nGRD with less than 3 occurrences:")
filtered_grd = filter(x -> x[2] <= 3, collect(grd_counts))
total_count_grd_less_3 = sum(x[2] for x in filtered_grd)
println("This is total of: $(length(filtered_grd)), and all of the cases are $total_count_grd_less_3:")
for (i, (grd, count)) in enumerate(filtered_grd)
    println("    $(i)) $grd -> $count times")
end
```

```julia
grd_menor_igual_3_porcentaje = round(total_count_grd_less_3 / rows * 100, digits=2)
println("Cantidad de enfermedades no comunes en el dataset: $(total_count_grd_less_3), y esto es un $(grd_menor_igual_3_porcentaje)% del dataset")
```

```julia
# Extraer solo los GRD con más de 3 apariciones
valid_grds = Set([grd for (grd, count) in grd_counts if count > 3])

# Filtrar el DataFrame original
df_filtered = filter(row -> row.GRD in valid_grds, df)

# Mostrar un resumen del nuevo DataFrame
println("Filtered DataFrame has $(nrow(df_filtered)) rows.")
```
