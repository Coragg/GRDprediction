```julia
function eliminar_columnas_con_pocos_datos(df::DataFrame, porcentaje_umbral::Float64 = 0.9)::DataFrame
    umbral = Int(ceil(porcentaje_umbral * nrow(df)))
    columnas_a_mantener = [col for col in names(df) if count(!ismissing, df[!, col]) >= umbral]
    return df[!, columnas_a_mantener]
end
```

```julia
clean_data = eliminar_columnas_vacias(filter_dataset, 0.10)
first(clean_data, 3)
```
