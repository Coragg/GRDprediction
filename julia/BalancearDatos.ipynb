```julia
function is_vacio(x)
    if ismissing(x) || x == nothing || x == ""
        return true
    elseif isa(x, Number) && isnan(x)
        return true
    else
        return false
    end
end

# Función para eliminar columnas con un umbral de valores vacíos
function eliminar_columnas_vacias(df::DataFrame, umbral::Float64 = 0.85)::DataFrame
    proporciones_vacias = [count(is_vacio, df[!, col]) / nrow(df) for col in names(df)]
    columnas_a_mantener = [col for (i, col) in enumerate(names(df)) if proporciones_vacias[i] < umbral]
    return df[!, columnas_a_mantener]
end
```

```julia
clean_data = eliminar_columnas_vacias(filter_dataset, 0.10)
first(clean_data, 3)
```

```julia
function count_vacios_as_dataframe(df::DataFrame)::DataFrame
    columnas = names(df)
    conteo = [count(is_vacio, df[!, col]) for col in columnas]
    return DataFrame(Columna=columnas, Vacios=conteo)
end
conteo_vacios = count_vacios_as_dataframe(clean_data)
```
