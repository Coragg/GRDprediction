```julia
using CSV, DataFrames, Imbalance, MLJ, MultivariateStats, CategoricalArrays 
```

```julia
file_path = "../data/filtered_data_elpino.csv"
df = CSV.read(file_path, DataFrame; missingstring=["NA", "NULL", "-"], delim=';')
first(df, 3)
```

```julia
names(df)  # Muestra todos los nombres de columnas
```

```julia
# Chequear el tipo de GRD
println(scitype(df."GRD"))  # Debe mostrar `AbstractVector{<:Finite}`

# Ver distribución por clase
combine(groupby(df, "GRD"), nrow => :count)
```

```julia
dropmissing!(df, :"GRD")  # Eliminar filas con GRD faltante
```

```julia
# 2. Convertir variables categóricas
df."GRD" = categorical(df."GRD")
df."Sexo (Desc)" = levelcode.(categorical(df."Sexo (Desc)"))  # Convertir sexo a numérico
```

```julia
# Verificar valores únicos en GRD
println("Valores únicos en GRD antes de limpiar: ", unique(df."GRD"))

# Eliminar filas donde GRD es missing o vacío
filter!(row -> !ismissing(row."GRD") && !isempty(string(row."GRD")), df)

# 2. Convertir a categórico con manejo seguro
try
    df."GRD" = categorical(df."GRD")
catch e
    @warn "Error al convertir GRD a categórico: $e"
    # Si falla, forzar tipo String primero
    df."GRD" = categorical(string.(df."GRD"))
end

# Verificar resultado
println(scitype(df."GRD"))  # Debería mostrar AbstractVector{<:Finite}
```
