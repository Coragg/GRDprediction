# Objetivo y funciones para las metricas

```python
def medir_modelos(real, predicho):
    accuracy = accuracy_score(real, predicho) * 100
    print(f"Accuracy: {accuracy:.2f}%")
    
    precision = precision_score(real, predicho, average='macro', zero_division=0) * 100
    print(f"Precision (macro): {precision:.2f}%")
    
    recall = recall_score(real, predicho, average='macro', zero_division=0) * 100
    print(f"Recall (macro): {recall:.2f}%")
```

```python
data_train.GRD = data_train.GRD.astype('category')
data_test.GRD = data_test.GRD.astype('category')
```

```python
global_target = data_train["GRD"]
global_target_test = data_test["GRD"]
```

```python
columns_name = data_train.columns.tolist()
columns_name.remove("GRD")
columns_name[0:4]
```

## Parametros para los modelos

```python
modelos = [["Diag 01 Principal (cod+des)", "Diag 02 Secundario (cod+des)", "Proced 01 Principal (cod+des)", "Proced 03 Secundario (cod+des)", "Proced 04 Secundario (cod+des)", "Proced 05 Secundario (cod+des)", "Edad en años", "Sexo (Desc)"], # Modelo 1
           ["Diag 01 Principal (cod+des)", "Diag 02 Secundario (cod+des)", "Proced 01 Principal (cod+des)", "Proced 03 Secundario (cod+des)", "Proced 04 Secundario (cod+des)", "Proced 05 Secundario (cod+des)", "Edad en años"], # Modelo 2
           ["Diag 01 Principal (cod+des)", "Diag 02 Secundario (cod+des)", "Proced 01 Principal (cod+des)", "Proced 03 Secundario (cod+des)", "Proced 04 Secundario (cod+des)", "Proced 05 Secundario (cod+des)", "Sexo (Desc)"] 
          ]
```

```python
modelos.append(columns_name)
```

## Modelo 1

```python
inputs_modelo1 = np.asanyarray(data_train[modelos[0]])
test_input_model1 = np.asanyarray(data_test[modelos[0]])
```

```python
scaler1 = StandardScaler().fit(inputs_modelo1)
x_train1 = scaler1.transform(inputs_modelo1)
x_test1 = scaler1.transform(test_input_model1)
```

### Modelo RBF

```python
modelo_rbf1 = svm.SVC(kernel='rbf')
modelo_rbf1.fit(x_train1, global_target)
```

```python
svm_prediction_rbf1 = modelo_rbf1.predict(x_test1)
print(svm_prediction_rbf1)
```

```python
medir_modelos(global_target_test, svm_prediction_rbf1)
```

### Modelo sigmoid

```python
modelo_sigmoid1 = svm.SVC(kernel='sigmoid')
modelo_sigmoid1.fit(x_train1, global_target)
```

```python
svm_prediction_sigmoid1 = modelo_sigmoid1.predict(x_test1)
print(svm_prediction_sigmoid1)
```

```python
medir_modelos(global_target_test, svm_prediction_sigmoid1)
```
